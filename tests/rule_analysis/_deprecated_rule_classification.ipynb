## Read in current Regression

```python
reg = pd.read_json("/Users/verisian/code/core-regression-dashboard/data/2025-11-20-rules.json")
print(reg['standard_source'].unique().tolist())
reg = reg[reg['standard_source']=='SDTMIG']
print(reg['standard_source'].unique().tolist())
print(reg.shape)
reg.head(200)
```

```python
def v_stat(row):
    if not pd.isna(row['error']):
        return 'error'
    if isinstance(row['negative_regressions'], list):
        response = []
        for tc in row['negative_regressions']:
            for k,v in tc.items():
                engine_regression = v.get("engine_regression")
                if engine_regression:
                    response.append(engine_regression.get("sql_overall_result"))
                else:
                    response.append('not success')
        if all(x == "success" for x in response):
            return 'success'
        else:
            return 'failing'
    return 'no test data'

reg["verisian_status"] = reg.apply(lambda row: v_stat(row), axis=1)
reg.head(200)
```

```python
def v_val(row):
    if not pd.isna(row['error']):
        return 'error'
    if isinstance(row['negative_regressions'], list):
        response = []
        for tc in row['negative_regressions']:
            for k,v in tc.items():
                engine_regression = v.get("engine_regression")
                if engine_regression:
                    response.append(engine_regression.get("sql_results_validation"))
                else:
                    return 'missing'
        if all(x == "success" for x in response):
            return 'valid'
        else:
            return 'invalid'
    return 'missing'
reg["verisian_validity"] = reg.apply(lambda row: v_val(row), axis=1)

def cg_id(row):
    if row['negative_folder_path']:
        return row['negative_folder_path'].split('/')[1]
    else:
        return ""

reg['negative_folder_path'] = reg['negative_folder_path'].fillna("")
reg['cg_rule_id'] = reg.apply(lambda row: cg_id(row), axis=1)

reg.head(200)
```

## merge regression

```python
reg = reg[['core-id', 'cdisc_rule_id', 'check_operators', 'verisian_status', 'verisian_validity', 'cg_rule_id']].copy()
merged = me.merge(reg, left_on='Rule Id', right_on='cg_rule_id', how='outer')
merged.head()
```

```python
merged.value_counts(["Rule Type"])
```

```python
merged["verisian_comment"] = ''
merged["verisian_status"] = merged["verisian_status"].astype(str)
merged["verisian_status"] = merged["verisian_status"].str.replace("nan", "not implemented")
merged['Rule Type'] = merged['Rule Type'].fillna("UNKNOWN")
merged.columns = [col.strip() for col in merged.columns]
merged.columns = [re.sub(r'[^A-Za-z0-9()]+', '_', col) for col in merged.columns]
merged.columns
```

```python
ordered = [
    'Rule_Id', 'Core_Id', 'Rule_ID', 'Rule_ID_Version', 'Rule_Identifier_Id', 'Rule_Identifier_Rule_Version',
    'Authority_Organization', 'Standards_Name', 'Standards_Version', 'Rule_Identifier_Conformance_Guide_Version',
    'Cited_Item', 'Cited_Guidance', 'Cited_Doc', 'Cited_Section', 'Release_Notes', 'Citations', 'References_Origin', 'Sensitivity', 'Severity',

    'verisian_comment', 'verisian_status', 'verisian_validity',
    'Description', 'Outcome_Message', 'Discussion_Notes', 'Rule_Type',
    'Variable', 'Condition', 'Rule',
    'Scopes_Classes', 'Scopes_Domains', 'Class', 'Subclass', 'Dataset_or_Domain',
    'Check', 'Match_Datasets', 'Operations',
    'Outcome_Output_Variables',

    'Related_Rule(s)', 'Rule_Set',

    'Jira_Executability', 'Team', 'Assigned_To', 'Update_Status',
    'CORE_Programmability_Status', 'CORE_Programmability_Comments',
    'Historical_Programmable', 'Historical_Programmable_Flag_Comment','Core_Version'
       ]
merged = merged[ordered].copy()
```

```python
merged.dtypes
```

```python
merged = merged.sort_values(by=["verisian_status"], ascending=True).copy()

date = str(datetime.today().date())
if not os.path.exists(date):
    os.mkdir(date)
for value, group in merged.groupby("Rule_Type"):
    cleaned = re.sub(r"[^A-Za-z0-9_-]+", "_", str(value))
    filename = f"{date}/{cleaned}.csv".lower()
    group.to_csv(filename, index=False, header=True)
```
